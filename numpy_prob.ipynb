```python
#finding dimensions of the arrays
import numpy as np
a=np.array(42)
b=np.array([2,6,4,9,10])
c=np.array([[10,20,30],[90,80,70]])
d=np.array([[[22,33,44],[12,13,14]],[[77,66,88],[23,98,67,]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

```python
#printing sum of elements inside the array
import numpy as np
arr=np.array([1,2,9,8,7])
print(arr[2]+arr[4])
```

```python
#printing the elemennt at the entered index
import numpy as np
arr=np.array([1,2,9,8,7])
print(arr[2])
```

```python
#accesing elements of 2-d array
import numpy as np
arr=np.array([[10,20,30],[90,80,70]])
print("3rd element of 2nd row:",arr[1,2])
```

```python
#slicing
import numpy as np
arr=np.array([2,6,4,9])
print(arr[1:4])
```

```python
#slicing
import numpy as np
arr=np.array([2,6,4,77,23,90])
print(arr[:5])
```

```python
#slicing
import numpy as np
arr=np.array([2,6,4,77,23,90])
print(arr[:])
```

```python
#negative slicing
import numpy as np
arr=np.array([1,2,3,4,5,6])
print(arr[-4:-1])
```

```python
#step slicing
import numpy as np
arr=np.array([1,2,3,4,5,6])
print(arr[1:3:2])
```

```python
#return every other
import numpy as np
arr=np.array([1,2,3,4,5,6])
print(arr[::3])
```
