```python
#to concatenate the 3 dictionaries into a new one

d1={1:10,2:20}
d2={3:90,4:80}
d3={5:67,6:38}
new_dic={}
new_dic.update(d1)
new_dic.update(d2)
new_dic.update(d3)
print("concatenated dictionary:",new_dic)
```

```python
#to check if a key exists in a dictionary

d={1:"mahak",2:"ayush",3:"gracy"}
key=int(input("Enter key to search:"))
if key in d:
    print("Key found")
else:
    print("Key not found")
```

```python
#to print keys alone,values alone and keys and values alone

d={
    1:"mahak",
    2:"ayush",
    3:"gracy"}
print("keys:")
for key in d:
 print("\nValues")
for value in d.values():
    print(value)
print("\nKeys and values:")
for key,value in d.items():
    print(key,value)
```

```python
#to sum all values in dictionary

d={"a":10,"b":55,"c":99}
t=0
for value in d.values():
    t+=value
print("sum:",t)
```

```python
#add ,delete and display

d={1:"hi",2:"bye",3:"hello"}
print("original dictionary:",d)
d[3]="good"
del d[2]
print("altered dictionary:",d)
```

```python
#convert list to dictionary
l1=["name","panel","no"]
l2=["mahak","a",2]
d=dict(zip(l1,l2))
print("dictionary:",d)
```

```python
#convert dictionary to list

d={1:"a",2:"c",3:"v"}
k_l=list(d.keys())
v_l=list(d.values())
print("list of keys:",k_l)
print("list of values:",v_l)
```

```python
#find mean of values

d={1:99,2:88,3:100,4:69}
t=sum(d.values())
mean=t/len(d)
print("Mean:",mean)
```
