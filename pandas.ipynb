```python
#creating series from list
import pandas as pd
a=[3,5,7]
myvar=pd.Series(a)
print(myvar)
```

```python
#naming your own labels
import pandas as pd
a=[3,5,7]
myvar=pd.Series(a,index=["x","y","z"])
print(myvar)
```

```python
#naming the series
s1=pd.Series([10,90,30],index=["a","b","c"],name="examples series")
print(s1)
```

```python
#series from dictionary
s2=pd.Series({"x":1,"y":2,"z":3})
print(s2)
```

```python
#dataframes
import pandas as pd
data={
    "calories":[234,567,789],
    "duration":[30,40,50]
}
myvar=pd.DataFrame(data)
print(myvar)
```

```python
#to access the value of the variable at 0 and 1th location
import pandas as pd
data={
    "calories":[234,567,789],
    "duration":[30,40,50]
}
myvar=pd.DataFrame(data)
print(myvar.loc[[0,1]])
```

```python
#data preprocessing
import pandas as pd
data={
    "a":[1,2,3,None,5],
    "b":[None,8,6,4,9],
    "c":[1,2,None,None,5]
}
df=pd.DataFrame(data)
print("original data:\n",df)
print()

#using dropna to remove rows with any missing values
df_clean=df.dropna()
print("cleaned data:\n",df_clean)
print()
```

```python
#how to fill values
import pandas as pd
data={
    "a":[1,2,3,None,5],
    "b":[None,8,6,4,9],
    "c":[1,2,None,None,5]
}
df=pd.DataFrame(data)

#filling 0 in place of none values
df.fillna(0,inplace=True)
print("original data:\n",df)
print()
```

```python
import pandas as pd
data={
    "a":[1,2,3,None,5],
    "b":[None,8,6,4,9],
    "c":[1,2,None,None,5]
}
df=pd.DataFrame(data)

df.rename(columns=({"a":"age","b":"name","c":"salary"}),inplace=True)
print(df.to_string(index=False))
```
