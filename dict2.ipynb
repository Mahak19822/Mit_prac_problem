```python
#to sort the keys in the dictionary
s={"name":"abc","no":90,"roll":23}
sorted_dict=sorted(s)
print("Sorted keys:",sorted_dict)
```

```python
#to print the key and value of the highest value
s={"Amit":45,"mahak":100,"ayush":65,"gracy":87}
h_s=max(s,key=s.get)
print("student with highest marks:",h_s)
print("highest mark of that student:",s[h_s])
```

```python
#to count frequency of each item in a list using dictionary
num=[91,28,5,6,77,28,43]
f={}
for n in num:
    if n in f:
        f[n]=f[n]+1
    else:
      f[n]=1
print("Frequency of elements:",f)
```

```python
#to sort a dictionary  by its values
s={"Amit":45,"mahak":100,"ayush":65,"gracy":87}
sorted_d=dict(sorted(s.items(),key=lambda item: item[1]))#in asc
print("ascending sorted dictionary:",sorted_d)
sorted_d=dict(sorted(s.items(),key=lambda item: item[1],reverse=True))#in des
print("descending sorted dictionary:",sorted_d)
```
