```python
#to print 4th element from the last in a tuple

num=(10,20,30,40,50,60,70,80,90)
print("Tuple:",num)
print("4th element from first:",num[3])
print("4th element from last:",num[-4])
```

```python
#to check if an element exists in tuple or not

num=(22,33,44,55,66)
ele=int(input("enter element to search"))
if ele in num:
    print("Element exists in tuple")
else:
    print("Element does not exist")
```

```python
#to find the index of an element in tuple

num=(10,20,30,40,50)
ele=int(input("Enter the element"))
if ele in num:
    print("Index of the element:",num.index(ele))
else:
    print("Element not found")
```

```python
#to convert a list into a tuple

num_list=[12,23,34,45,56]
num_tup=tuple(num_list)
print("list:",num_list)
print("Tuple:",num_tup)
```

```python
#to replace last values in tuple

num=[(10,20,40),(40,50,60),(70,80,90)]
new_list=[]
for tup in num:
    new_tup=tup[:-1]+(100,)
    new_list.append(new_tup)
print("original list:",num)
print("updated list:",new_list)
```

```python
#to find max and min element

num=[22,31,42,67,89]
maximum=max(num)
minimum=min(num)
print("Minimum element:",minimum)
print("maximum element:",maximum)
```

```python
#to swap first and last element in tuple

num=(55,43,45,76)
new_tuple=(num[-1],)+num[1:-1]+(num[0],)
print("original list:",num)
print("after swapping:",new_tuple)
```
